```python
df_clean = df.copy()

df_clean = df_clean.drop_duplicates()

df_clean["Installs"] = (
    df_clean["Installs"]
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
    .astype(int)
)

df_clean["Price"] = (
    df_clean["Price"]
    .str.replace("$", "", regex=False)
    .astype(float)
)

df_clean["Last Updated"] = pd.to_datetime(
    df_clean["Last Updated"],
    errors="coerce"
)
```

```python
df_clean.dtypes
```

```python
df_clean.shape
```

```python

df_clean["Size_MB"] = (
    df_clean["Size"]
    .str.replace("M", "", regex=False)
    .replace("Varies with device", np.nan)
)


df_clean["Size_MB"] = pd.to_numeric(
    df_clean["Size_MB"],
    errors="coerce"
)
```

```python
missing_percentage = (
    df_clean.isnull().sum() / len(df_clean) * 100
).round(2)

missing_percentage
```

```python
df_rating = df_clean.dropna(subset=["Rating"]).copy()
```

```python
df_rating.shape
```

```python
df_rating["Rating"].isnull().sum()
```

```python
df_clean["Is_Paid"] = df_clean["Type"].map({
    "Free": 0,
    "Paid": 1
})
```

```python
df_clean[["Type", "Is_Paid"]].head(10)
```

```python
def install_range(installs):
    if installs < 10000:
        return "Low"
    elif installs < 1000000:
        return "Medium"
    elif installs < 10000000:
        return "High"
    else:
        return "Very High"

df_clean["Install_Range"] = df_clean["Installs"].apply(install_range)
```

```python
df_clean["Install_Range"].value_counts()
```

```python
df_clean["Review_Rate"] = (
    df_clean["Reviews"] / df_clean["Installs"] * 100
)
```

```python
df_clean["Review_Rate"].describe()
```

```python
reference_date = pd.Timestamp("2018-12-31")

df_clean["App_Age_Years"] = (
    (reference_date - df_clean["Last Updated"]).dt.days / 365.25
).round(2)
```

```python
df_clean[["Last Updated", "App_Age_Years"]].head()
```

```python
def price_category(price):
    if price == 0:
        return "Free"
    elif price <= 5:
        return "Low"
    elif price <= 20:
        return "Medium"
    else:
        return "High"

df_clean["Price_Category"] = df_clean["Price"].apply(price_category)
```

Which categories have the most installs?

```python
category_performance = (
    df_clean.groupby("Category")
    .agg(
        Total_Apps=("App", "count"),
        Total_Installs=("Installs", "sum"),
        Average_Rating=("Rating", "mean"),
        Total_Reviews=("Reviews", "sum")
    )
    .sort_values("Total_Installs", ascending=False)
)

category_performance.head(10)
```

```python
plt.figure(figsize=(12, 6))

category_performance["Total_Installs"].head(10).sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Categories by Total Installs")
plt.xlabel("Total Installs")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```

Which categories have the highest average rating?


```python
category_ratings = (
    df_clean.groupby("Category")["Rating"]
    .mean()
    .sort_values(ascending=False)
)

category_ratings.head(10)
```

```python
plt.figure(figsize=(12, 6))

category_ratings.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Categories by Average Rating")
plt.xlabel("Average Rating")
plt.ylabel("Category")
plt.xlim(3.5, 5)

plt.tight_layout()
plt.show()
```
